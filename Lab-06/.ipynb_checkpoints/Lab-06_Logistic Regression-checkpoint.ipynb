## Filling null values of age column

```python
df.marital_status= df.marital_status.fillna(handle)
```

```python
x= df[['age']]
```

```python
y= df['marital_status']
```

##  Spliting dataset for train and test

```python
from sklearn.model_selection import train_test_split
```

```python
#xtrain,ytrain,xtest,ytest = train_test_split(x,y,test_size= .20, random_state= 1)
xtrain, xtest, ytrain, ytest = train_test_split(x, y, test_size = 0.30, random_state =1)
```

```python
xtest
```

```python
ytest
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression ()
```

##  Model Training 

```python
model.fit(xtrain, ytrain)
```

```python
model.predict(xtest)
```

```python
#Finding out accuracy
model.score(xtest,ytest)
```

```python
#predicting performance using sigmoid function
model.predict_proba(xtest)
# here first values in the array is for no, second value is for yes
```

```python
# Predicting status of anyother single value providing age (one of the ways)
# Assuming you have a trained logistic regression model called 'model'
# model = LogisticRegression() # Your trained model here
def predict_marriage_status(model, age):
    
    # Prepare the input as a 2D array
    age_input = np.array([[age]])
    
    # Get the prediction: 1 for married (Yes), 0 for not married (No)
    prediction = model.predict(age_input)[0]
    
    # Convert prediction to readable format
    return "Yes" if prediction == 1 else "No"

# Example usage
age = 42 # Replace with any age you want to predict
status = predict_marriage_status(model, age)

print(f"Prediction: {status}")
```

```python
# Predicting status of anyother single value providing age (second another way)
age = 30 # Replace with any age you want to predict
status = "Yes" if model.predict([[age]])[0] == 1 else "NO"

print(f"Prediction: {status}")
```

```python
# Predicting status of anyother single value providing age (third another way)
# Assuming you have a trained logistic regression model called 'model'
# model = LogisticRegression() # Your trained model here

# Predict marriage status for a given age
age = 30 # Replace with any age you want to predict
status = ["No", "Yes"] [int(model.predict([[age]])[0])]

print(f"Prediction: {status}")
```
